```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt


df = pd.read_csv("./Income_Data.csv")

df.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
import numpy as np 

def ds(column): 
    min = np.min(df[column])
    first_Q = np.percentile(df[column], 25)
    med = np.median(df[column])
    third_Q = np.percentile(df[column], 75)
    max = np.max(df[column])
    mean = np.mean(df[column])

    print (f"{column} Data Number Summary")
    print (f"Min: {min}")
    print (f"Q1: {first_Q:,.2f}")
    print (f"Median: {med:,.2f}")
    print (f"Q3: {third_Q:,.2f}")
    print (f"Max: {max}")
    print (f"Mean: {mean:,.2f}")

    print ()

ds ("income")
ds ("education")
ds("prestige")
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
# Code here

standard_dev = df['income'].std() 
variance = standard_dev ** 2

print (f"Standard deviation of Income: {standard_dev:,.2f}")
print (f"Variance of Income: {variance:,.2f}")
```
