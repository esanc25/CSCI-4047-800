```python
import pandas as pd
import scipy.stats as stats

#inspect data 
df = pd.read_csv("./Americandata.csv")
df.head(10)
```

```python
# Contingency tables
def contingency_table (name):
    #create contingency table 
    table = pd.crosstab (df["Happy"], df[name])
    print(name.upper())
    print(f"Contingency table:\n {table}")

    #run chi-square testing
    chi2, p_value, dof, expected = stats.chi2_contingency(table, correction=False)
    print(f"\nChi-square test: \nChi-square: {chi2}\np_value: {p_value:.6f}\ndegrees of freedom: {dof}\n")

    #return expected outcomes 
    print("Expected counts- if variables were UNRELATED:")
    expected_df = pd.DataFrame(expected, index = table.index, columns = table.columns)
    print(expected_df.round(2))
    #checking assumption 
    if (expected >=5).all():
        print("Assumption okay; all counts greater than 5.")
    else:
        print("WARNING! some expected counts may be less than 5.")

    print()

    #critical values decison
    alpha = 0.05
    critical_value = stats.chi2.ppf(1 - alpha, dof)
    print(f"Critical value (alpha = {alpha}. Degree of freedom = {dof}): {critical_value:.2f}")
    print(F"Our Chi-square= {chi2}")


    #p-value decision
    if chi2 > critical_value:
        print(f"Since {chi2:.2f} > {critical_value:.2f}, Reject independence (reject the null hypothesis) -> role and hero ARE related")
    else:
        print(f"Since {chi2:.2f} <= {critical_value:.2f}, Cannot reject") #you accept null, there is no relation

    print()

    if p_value <= alpha: 
        print(f"{p_value:6f} < {alpha}: Reject null hypothesis- happiness and profession are related \n(dependent)")
    else: 
        print(f"{p_value:6f} >= {alpha}: Fail to reject null hypothesis- there is not enough evidence that happiness and profession are related\n(independent)")

    print("="*120)
    print()


contingency_table("Students")
contingency_table("Laborers")
contingency_table("Preachers")
contingency_table("Physicians")
contingency_table("Housewives")
contingency_table("Teachers")
contingency_table("Lawyers")
contingency_table("Musicians")
```
